```python

import numpy as np
```

```python
a=np.array([1,2,3])
print("Vector",a)
```

```python
b=np.array([[1,2,3],[1,2,3]])
print("Matrix",b)
```

```python
c=np.array([
    [[1,2,3],[1,2,3]],
    [[1,2,3],[1,2,3]]
])
print("Tensor")
print(c)
```

```python
#different types of numpy arrays
float_arr=np.array([11,13],dtype=float)
print(float_arr)

bool_arr=np.array([0,1,3],dtype=bool)
print(bool_arr)
```

```python
#np.arange
a=np.arange(1,11)
print(a)
b=np.arange(1,15,3)
print(b)
```
