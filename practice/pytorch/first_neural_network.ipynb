```python
import torch
import torch.nn as nn
```

```python
x=torch.tensor(3.0)
target=torch.tensor(10.0)

w=torch.tensor(2.0,requires_grad=True)

prediction=x*w

loss=(prediction-target)**2
loss.backward()

print(w.grad)
```

```python
x=torch.tensor(3.0)
target=torch.tensor(10.0)
w=torch.tensor(2.0,requires_grad=True)
learning_rate=0.1


for epoch in range(10):
    prediction=w*x
    loss=(prediction-target)**2
    loss.backward()
    with torch.no_grad():
        w-=learning_rate*w.grad
    print(w.grad)
    w.grad.zero_()
```
