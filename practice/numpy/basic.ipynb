```python

import numpy as np
```

```python
a=np.array([1,2,3])
print("Vector",a)
```

```python
b=np.array([[1,2,3],[1,2,3]])
print("Matrix",b)
```

```python
c=np.array([
    [[1,2,3],[1,2,3]],
    [[1,2,3],[1,2,3]]
])
print("Tensor")
print(c)
```

```python
#different types of numpy arrays
float_arr=np.array([11,13],dtype=float)
print(float_arr)

bool_arr=np.array([0,1,3],dtype=bool)
print(bool_arr)
```

```python
#np.arange
a=np.arange(1,11)
print(a)
b=np.arange(1,15,3)
print(b)
```

```python
#changing dimensions
reshaped_array=np.arange(1,11).reshape(5,2)
print(reshaped_array)
```

```python
random_array=np.random.random((3,4))
print(random_array)
```

```python
ones_matrix=np.ones((2,1))
print(ones_matrix)
zeros_matrix=np.zeros((4,5))
print(zeros_matrix)
```

```python
linearly_space=np.linspace(1,18,15)
print(linearly_space)

#okay so this creates an even spaced matrix from 1 to 18 having limit of 15 numbers we can change the limit
```

```python
identity_matrix=np.identity(3)
print(identity_matrix)
```

```python
a=np.arange(1,11)
b=np.arange(1,11)
print(a *2)
print(a+2)
print(a>5)
print(a==9)
```

```python
result_add=a+b 
print(result_add)
result_multi=a*b 
print(result_multi)
```

```python
##Mathematical and Statistical Methods
print(np.max(a))
np.min(b)
np.sum(a,axis=0)#0means along the rows

print(np.prod(a))

print(np.mean(b))
print(np.median(a))
print(np.std(a))
print(np.var(b))

np.sin(a)
np.cos(b)

# Dot Product (Matrix Multiplication) [37, 38] # Requires column count of 1st matrix == row count of 2nd matrix [37] 
print(np.dot(a, b))

print(np.log(a))
print(np.exp(b))
print(np.round(a))
print(np.floor(a))
print(np.ceil(b))
```
